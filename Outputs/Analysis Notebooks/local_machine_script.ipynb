```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/Users/haydenmonk/Downloads/HPCC/Mass_Fraction/Outputs/Ejection_Results/local_20260726_225705_ejection_results.txt', header=None)
```

```python
df.columns = ['m_planet','t', 'bash_id', 'result']
```

```python
grouped = (
    df.groupby(["m_planet", "t"])
      .agg(
          total_particles=("result", "size"),
          ejected_particles=("result", lambda x: x.eq("Ejected").sum())
      )
      .reset_index()
)

grouped["ejection_fraction"] = (
    grouped["ejected_particles"] / grouped["total_particles"]
)
```
